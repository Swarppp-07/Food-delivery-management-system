```python
class user:
  def __init__(self,name,user_id):
    self.name=name
    self.user_id=user_id

  def display_details(self):
    print("name :",self.name)
    print("user_id :",self.user_id)

#class customer and inherets from user
class customer(user):
  def __init__(self, name, user_id, address):
    super().__init__(name, user_id)
    self.address = address


  def place_order(self, order_obj):
        print(f"\nOrder {order_obj.order_id} Placed Successfully by {self.name}!")


#class premimum customer inherets from customer

class premiumcustomer(customer):
  def __init__(self,name,user_id,address):
    super().__init__(name, user_id, address)

  def calculate_discount(self):

    return 0.20


#class restaurant to sotre restarunt info
class restaurant:
  def __init__(self, id, restaurant_name, location):
    self.id = id
    self.restaurant_name=restaurant_name
    self.location = location

  def display_details(self):
    print("Restaurant ID :", self.id)
    print("Restaurant Name :",self.restaurant_name)
    print("Location :", self.location)


#class delivery partner with details

class deliverypartner:
  def __init__(self,name,partner_id):
    self.name=name
    self.partner_id=partner_id
    self.available = True
  def display_details(self):
    print("Delivery partner is assigned :",self.name)
    print("Partner ID :", self.partner_id)

#class order

class order:
  def __init__(self, order_id, customer, restaurant, items):
    self.order_id = order_id
    self.customer = customer
    self.restaurant = restaurant
    self.items = items
    self.partner = None
    self.delivery_time = 30


  def calculate_discount(self):
      total_amount = sum(price * qty for _, price, qty in self.items)

      if isinstance(self.customer, premiumcustomer):

          return total_amount * self.customer.calculate_discount() if hasattr(self.customer, 'calculate_discount') else total_amount * 0.20
      return total_amount * 0.10

  def assign_delivery(self, partner):
      if partner.available:
          self.partner = partner
          partner.available = False
          print(f"Delivery Partner: {self.partner.name} ({self.partner.partner_id}) assigned for Order {self.order_id}")
      else:
          print(f"Delivery Partner: {partner.name} is currently unavailable for Order {self.order_id}")


  def generate_bill(self):
      subtotal = sum(price * qty for _, price, qty in self.items)
      discount = self.calculate_discount()
      delivery_charge = 40
      total = subtotal - discount + delivery_charge

      print("\n--- BILL ---")
      print("Order ID:", self.order_id)
      print("Customer:", self.customer.name)
      print("Restaurant:", self.restaurant.restaurant_name)
      print("Items:")
      for item_name, price, qty in self.items:
          print(f"  - {item_name} (₹{price} x {qty})")
      print(f"Subtotal: ₹{subtotal:.2f}")
      print(f"Discount: ₹{discount:.2f}")
      print(f"Delivery: ₹{delivery_charge:.2f}")
      print(f"Total: ₹{total:.2f}")
      if self.partner:
          print(f"Delivery Partner: {self.partner.name}")
      print(f"Estimated Delivery Time: {self.delivery_time} minutes")
      print("----------------\n")


  def display_details(self):
      print("\nOrder ID:", self.order_id)
      print("Restaurant:", self.restaurant.restaurant_name)
      print("Customer:", self.customer.name)
      print("Items:")
      for item_name, price, qty in self.items:
          print(f"  - {item_name} (₹{price} x {qty})")
      if self.partner:
          print("Delivery Partner:", self.partner.name)
      else:
          print("Delivery Partner: Not yet assigned")



customer1 = customer("Rahul", 101, "9876543210")
customer2 = premiumcustomer("Nikhil", 102, "9876501234")

restaurant1 = restaurant(1, "Meghana Foods", "Bangalore")
restaurant2 = restaurant(2, "Empire Restaurant", "Bangalore")

partner1 = deliverypartner("Ravi", 201)
partner2 = deliverypartner("Kiran", 202)



order1 = order(
    1001, customer1, restaurant1,
    [('Biryani', 250, 2), ('Coke', 50, 2)]
)

customer1.place_order(order1)
order1.assign_delivery(partner1)
order1.display_details()
order1.generate_bill()

order2 = order(
    1002, customer2, restaurant2,
    [('Pizza', 300, 1), ('Garlic Bread', 150, 1)]
)

customer2.place_order(order2)
order2.assign_delivery(partner2)
order2.generate_bill()
order2.display_details()


print("\n--- CUSTOMER DETAILS ---")
customer1.display_details()
customer2.display_details()

print("\n--- RESTAURANTS ---")
restaurant1.display_details()
restaurant2.display_details()

print("\n--- DELIVERY PARTNERS ---")
partner1.display_details()
partner2.display_details()
```
